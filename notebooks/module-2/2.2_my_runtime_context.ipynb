```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from dataclasses import dataclass

@dataclass
class ColourContext:
    favourite_colour: str = "blue"
    least_favourite_colour: str = "pink"
```

```python
from langchain.tools import tool, ToolRuntime

@tool
def get_favourite_color(runtime: ToolRuntime):
    """Get the favourite colour of the user"""
    return runtime.context.favourite_colour

@tool
def get_least_favourite_color(runtime: ToolRuntime):
    """Get the least favourite colour of the user"""
    return runtime.context.least_favourite_colour
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model = "gpt-5-nano",
    context_schema=ColourContext,
    tools=[get_favourite_color, get_least_favourite_color]
)
```

```python
from langchain.messages import HumanMessage

message = HumanMessage(content="What is my favourite colour?")
```

```python
from pprint import pprint

response = agent.invoke(
    {"messages": [message]},
    context=ColourContext()
)

pprint(response)
```
